This notebook is generated from the governed Systems recipe. It does not install a model or touch Nexus production.

```python
import json, platform, shutil, subprocess, sys, time
def probe(command):
    p = subprocess.run(command, capture_output=True, text=True, timeout=30, check=False)
    return {'returncode': p.returncode, 'stdout': p.stdout[-2000:], 'stderr': p.stderr[-2000:]}
report = {'python': sys.version, 'platform': platform.platform(), 'started_at': time.time()}
report['torch'] = probe([sys.executable, '-c', 'import torch; print(torch.__version__); print(torch.cuda.is_available())'])
report['ffmpeg'] = probe(['ffmpeg', '-version'])
report['disk'] = shutil.disk_usage('/')._asdict()
json.dump(report, open('diagnostic_report.json', 'w'), indent=2, sort_keys=True)
json.dump({'status': 'SUCCEEDED', 'recipe': 'gpu-environment-diagnostic-v1', 'production_touch': False}, open('run_manifest.json', 'w'), indent=2, sort_keys=True)
```
